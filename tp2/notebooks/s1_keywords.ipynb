## Imports

```python
import os, subprocess, textwrap, sys, pandas as pd 
import yake
kw_extractor = yake.KeywordExtractor(lan="fr", top=50)
```

## Extraire les mots clés d'un document avec Yake

https://github.com/LIAAD/yake

```python
# Instantier l'extracteur de mots clés
from pathlib import Path, PurePath
import os

cwd = Path.cwd()
print("CWD =", cwd)
print("Ici →", os.listdir(".")[:10])
print("Parent →", cwd.parent, "|", os.listdir(cwd.parent)[:10])
```

```python
# Lister les Fichiers
from pathlib import Path
import os

data_path = (Path.cwd().parents[1] / "data" / "txt" / "1950").resolve()
print("data_path =", data_path)
assert data_path.exists(), f"Introuvable : {data_path}"

files = [f for f in os.listdir(data_path) if f.endswith(".txt")]
len(files)
```

```python
# Imprimer le nombre de fichiers identifiés
len(files)
```

```python

import csv
from pathlib import Path

OUT_DIR = (Path.cwd().parent / "output" / "keywords")   
OUT_DIR.mkdir(parents=True, exist_ok=True)
OUT_CSV = OUT_DIR / "keywords_1950.csv"

rows = []
for f in sorted(files):
    with open(data_path / f, "r", encoding="utf-8", errors="ignore") as fh:
        text = fh.read()
    kws = kw_extractor.extract_keywords(text)
    for kw, score in kws:
        if len(kw.split()) == 2:           
            rows.append([f, kw, score])

with open(OUT_CSV, "w", newline="", encoding="utf-8") as out:
    w = csv.writer(out)
    w.writerow(["doc_id", "keyword", "score"])
    w.writerows(rows)

print(f"OK → {OUT_CSV} | lignes écrites : {len(rows)}")
```

## Faire la même opération sur tous les documents

```python
for f in sorted(files)[:10]:
    text = open(os.path.join(data_path, f), 'r', encoding="utf-8").read()
    keywords = kw_extractor.extract_keywords(text)
    kept = []
    for kw, score in keywords:
        words = kw.split()
        if len(words) == 2:
            kept.append(kw)
    print(f"{f} mentions these keywords: {', '.join(kept)}...")
```
