### 3) Garde-fous: vérifier que le corpus est bien chargé

Avant d'aller plus loin, je mets des assertions pour m'assurer que le dossier data/txt/1950/ existe bien, qu'il contient effectivement des fichiers .txt, que le sous-ensemble "1950" n'est pas vide (et si besoin, que les noms de fichiers correspondent bien à 1950). Ces vérifications évitent de continuer avec un corpus mal monté ou un mauvais chemain, ce qui pourrait fausser les étapes suivantes. 

```python
from pathlib import Path

year_dir = Path("../data/txt/1950")

assert year_dir.exists(), f"Dossier introuvable : {year_dir}"
assert year_dir.is_dir(), f"Ce n'est pas un dossier : {year_dir}"

files_1950 = sorted(year_dir.glob("*.txt"))
assert len(files_1950) > 0, f"Aucun fichier .txt trouvé dans : {year_dir}"

bad = [p.name for p in files_1950 if "1950" not in p.name]
assert len(bad) == 0, f"Fichiers inattendus (pas '1950' dans le nom) : {bad[:5]}"

print("Garde-fous OK :", len(files_1950), "fichiers .txt dans", year_dir)
```

### 4) Export: enregistrer la liste des fichiers 1950 en JSON

Ici, j'exporte la liste complète des fichiers .txt de 1950 dans un fichier JSON (dans TP4/out/). L'intérêt est de pouvoir réutiliser exactement la même liste dans les notebooks/étapes suivantes (S2-S3), sans relister le dossier à chaque fois. Je garde les chemins sous forme de chaînes de caractères pour que ce soit simple à recharger ensuite. 

```python
from pathlib import Path
import json

year_dir = Path("../data/txt/1950")
files_1950 = sorted(year_dir.glob("*.txt"))

out_dir = Path("TP4/out")
out_dir.mkdir(parents=True, exist_ok=True)

json_path = out_dir / "files_1950.json"
data = [str(p.as_posix()) for p in files_1950]  

with open(json_path, "w", encoding="utf-8") as f:
    json.dump(data, f, ensure_ascii=False, indent=2)

print("JSON enregistré :", json_path, "|", len(data), "fichiers")
```
